```python
import numpy as np
import matplotlib.pyplot as plt
import soundfile as sf

from scipy.io.wavfile import write
from IPython.display import Audio, display
```

```python
frequency = 440

sampling_rate = 44100

duration = 5

amplitude = 0.5

time = np.arange(
    0,
    duration,
    1 / sampling_rate
)

audio_signal = amplitude * np.sin(
    2 * np.pi * frequency * time
)
```

```python
print("Number of samples:", len(audio_signal))

print("Sampling rate:", sampling_rate)

print("Duration:", len(audio_signal) / sampling_rate)

print("Maximum amplitude:", np.max(audio_signal))

print("Minimum amplitude:", np.min(audio_signal))
```

```python
plt.figure(figsize=(12, 4))

plt.plot(time[:1000], audio_signal[:1000])

plt.title("Digital Audio Waveform")

plt.xlabel("Time (seconds)")

plt.ylabel("Amplitude")

plt.grid(True)

plt.show()
```

```python
display(
    Audio(
        audio_signal,
        rate=sampling_rate
    )
)
```

```python
import os

os.makedirs("data/synthetic", exist_ok=True)

sf.write(
    "data/synthetic/440hz_tone.wav",
    audio_signal,
    sampling_rate,
    subtype="PCM_16"
)
```

```python
loaded_audio, loaded_sr = sf.read(
    "data/synthetic/440hz_tone.wav"
)

print("Sampling Rate:", loaded_sr)

print("Audio Shape:", loaded_audio.shape)

print("Duration:", len(loaded_audio) / loaded_sr)

print("Data Type:", loaded_audio.dtype)
```
